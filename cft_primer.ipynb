# A conformal field theory primer for the temporal-entropy method

*This chapter is a companion to the methods sections (block power method, `blockpm_methods`) and to
every results notebook that quotes a "CFT prediction". Its only job is to make the handful of
conformal-field-theory formulas we keep using — the central charge $c$, the transfer-matrix
spectrum, the boundary exponent $x_1$, and the entropy chord $\tfrac{c}{6}\log[\tfrac{2T}{\pi}\sin\tfrac{\pi t}{T}]$
— feel earned rather than quoted. It assumes no prior CFT, builds the ideas in order, and ends with
a **decoder ring** (§9) that reads each thesis formula piece by piece, plus a one-page glossary (§11).*

**The one-paragraph version.** At a critical point a lattice model forgets its microscopic details
and is governed by a *conformal field theory* — a continuum theory with an enormous symmetry that,
in two dimensions, pins down almost everything in terms of a single number, the **central charge**
$c$. Our whole method is an elaborate way of measuring $c$ (and a few companion numbers) from the
*temporal* direction of a quench. The reason it works is a beautiful accident of Euclidean physics:
after a Wick rotation, *time and space play interchangeable roles*, so the same CFT formulas that
describe a spatial slice also describe a temporal one — and the eigenvalues of our transfer matrix
are, quite literally, the CFT's list of operators. Everything below unpacks that sentence.

## 1. Why a field theory takes over at a critical point

Start with the physical picture, because it is what licenses everything else. Away from a phase
transition, a spin chain has a finite **correlation length** $\xi$: two spins farther apart than
$\xi$ are essentially independent. Microscopic details (the exact couplings, the lattice spacing
$a$) matter, because you can resolve them at the scale $\xi$.

At a continuous critical point $\xi \to \infty$. There is now *no* intrinsic length scale between
the lattice spacing $a$ and the system size $L$. A system with no scale looks the same at every
magnification — it is **scale invariant**. And here is the key consequence, the engine of
*universality*: if you zoom out, blurring the lattice into a continuum, the microscopic details wash
out and what survives is a continuum field theory that depends only on the *symmetries* and
*dimensionality* of the problem, not on its microscopic realisation. Wildly different microscopic
models — the Ising chain, a lattice gas, the Alcaraz ANNNI-type chain at small frustration — flow to
the *same* continuum theory. They are in the same **universality class**.

That continuum theory, at a critical point, is massless (no scale ⇒ no mass gap ⇒ gapless
excitations) and turns out to have a much larger symmetry than plain scale invariance. Promoting
"scale invariance + locality" to its full local form gives **conformal invariance**, and the theory
is a **conformal field theory (CFT)**. This is why we are allowed to replace the messy lattice
Hamiltonian by a clean CFT when we want to predict universal quantities — and why a number computed
from the Ising CFT can be compared directly against numbers extracted from our Alcaraz simulations.

## 2. What "conformal" means, and why 2D is special

A **conformal transformation** is a coordinate change that preserves *angles* but not necessarily
lengths — it can stretch, but only locally-isotropically (the same stretch in every direction at a
given point). Rotations, translations, and uniform rescalings are all conformal; so is any smoother
warping that locally looks like "rotate-and-scale".

In a generic number of dimensions the conformal group is finite-dimensional and only mildly
constraining. **Two dimensions is the miracle.** In 2D, "locally rotate-and-scale" is exactly what a
*holomorphic (analytic) function of a complex variable* does — multiplication by $f'(z)$ is a local
rotation-and-scale. So in 2D *every* analytic change of coordinates $z \mapsto f(z)$ is a conformal
symmetry, and there are infinitely many of them. The symmetry algebra becomes
**infinite-dimensional** (the Virasoro algebra, §5), and it constrains the theory so tightly that
2D CFTs are essentially solvable.

Two habits of thought come from this and we will use both:

- **Work on the complex plane.** Package the two Euclidean coordinates $(x,\tau)$ into $z = x + i\tau$
  and $\bar z = x - i\tau$. Physical quantities factorise into a *holomorphic* piece (depending on
  $z$) and an *antiholomorphic* piece (depending on $\bar z$). The paper's formulas with an $h$ and
  a separate $\bar h$ are exactly this split.
- **Change geometry by changing coordinates.** Because analytic maps are symmetries, if you know a
  correlation function on the plane you know it on *any* shape an analytic map can reach — a cylinder,
  a strip, a disk — just by substituting $z = z(\omega)$. This "conformal map" trick is how the paper
  gets from textbook plane results to our specific strip geometry (§7), and it is doing all the real
  work behind Eqs. (14), (17), (18).

*A note that will matter a lot later:* in **Euclidean** CFT the two coordinates $x$ and $\tau$ enter
completely symmetrically — nothing distinguishes "space" from "(imaginary) time". That democracy is
what makes the *transverse* / temporal method legitimate (§8): we are free to call whichever
direction we like "space".

## 3. The central charge $c$ — the one number to rule them all

Every 2D CFT carries a single most-important number, the **central charge** $c$. Several ways to
read it, all true at once:

- **It labels the universality class.** Fixing $c$ (together with the operator content) essentially
  fixes the theory. The models in this thesis and their central charges:

  | model | universality class | central charge $c$ |
  |---|---|---|
  | transverse-field Ising, **Alcaraz** ANNNI-type at small $p$ | Ising | $\tfrac{1}{2}$ |
  | tricritical Ising (O'Brien–Fendley point) | tricritical Ising | $\tfrac{7}{10}$ |
  | 3-state Potts | 3-state Potts | $\tfrac{4}{5}$ |
  | **XXZ** chain, $|\Delta|\le 1$ | free boson / Luttinger liquid | $1$ |

  The whole thesis question "does temporal Ising universality survive NNN frustration?" is, concretely,
  "does the Alcaraz chain still give $c=\tfrac12$ when measured through *time*?"

- **It is a "conformal anomaly" — the price of curving space.** Scale invariance is slightly broken
  by the geometry itself: putting the theory on a finite or curved region costs a universal
  ground-state energy proportional to $c$. This is the **Casimir energy**, and it is precisely how $c$
  will enter the transfer-matrix eigenvalues in §5 (the $-c/24$ term). This is also the most direct
  handle we have: *measure the finite-size ground-state energy or entropy, read off $c$.* Notebook 4
  does exactly this in equilibrium (DMRG central charge).

- **It counts gapless degrees of freedom.** Loosely, $c$ is proportional to the number of independent
  massless modes: $c=\tfrac12$ is "half" a degree of freedom (a single Majorana/Ising mode), $c=1$ is
  one full bosonic mode (the XXZ compact boson). Bigger $c$ ⇒ more entanglement, more computational
  cost — which is why the tricritical $c=\tfrac{7}{10}$ walls our method fastest.

Keep $c$ in mind as **the** target: nearly every formula below is ultimately a way to isolate it.

## 4. Operators, scaling dimensions, and the spectrum bridge

A CFT is organised not around particles but around **local operators** (also called *fields* or
*scaling operators*). The special ones, the **primary operators**, each carry two numbers, the
**conformal weights** $(h, \bar h)$, whose sum

$$ x = h + \bar h \tag{C1}$$

is the **scaling dimension** — how the operator responds to a rescaling. Scale invariance forces
correlation functions to be pure power laws set by these dimensions; on the infinite plane the
two-point function of a primary $\phi$ is (paper Eq. 12)

$$ \langle \phi(z_1)\phi(z_2)\rangle = \frac{1}{(z_1 - z_2)^{2h}\,(\bar z_1 - \bar z_2)^{2\bar h}} . \tag{C2}$$

The exponent *is* the scaling dimension. A small dimension $x$ = a slowly-decaying correlation = a
"light", easily-excited operator; a large $x$ decays fast and is "heavy".

Under a conformal map $\omega = f(z)$ a primary transforms by simple Jacobian factors (paper Eq. 13),
$\langle\phi(\omega_1)\phi(\omega_2)\rangle = \prod_i f'(z_i)^{-h}\,\bar f'(\bar z_i)^{-\bar h}\,
\langle\phi(z_1)\phi(z_2)\rangle$. Feeding in the plane→cylinder map $\omega = \tfrac{\beta}{2\pi}\log z$
turns the plane power law into the cylinder's hyperbolic-sine form (paper Eq. 14),

$$ \langle \phi(\omega_1)\phi(\omega_2)\rangle \;\propto\;
   \Big[\sinh\tfrac{\pi(\omega_1-\omega_2)}{\beta}\Big]^{-2h}\Big[\cdots\Big]^{-2\bar h}. \tag{C3}$$

We will meet the **time-like** version of exactly this (a $\sin$ instead of $\sinh$) as the source of
the entropy chord in §8. For now the headline is a dictionary we will use constantly:

> **The list of primary operators, with their dimensions $x$, is a fixed fingerprint of the CFT —
> and (next section) it is *literally* the list of eigenvalues of our transfer matrix.**

That correspondence — *operators $\leftrightarrow$ transfer-matrix eigenvalues* — is the single most
important bridge in the method. It is why staring at the spectrum of a matrix tells us about a
continuum field theory.

## 5. The stress tensor, $L_0$, and how $c$ enters the eigenvalues

We need to see *mechanically* how the abstract central charge $c$ turns into a number we can fit. The
carrier is the **stress–energy tensor** $T_{\mu\nu}$ — the operator measuring the theory's response to
a change of coordinates (paper Eq. 11). Because coordinate changes *are* the symmetry in 2D, the
stress tensor is the generator of conformal transformations, and its modes generate the
infinite-dimensional **Virasoro algebra**. One mode is special: $L_0$, the generator of scaling. Its
eigenvalues on the primary operators are their weights $h$ — so **$L_0$ is the "dictionary operator"
whose spectrum is the operator content.**

Now quantise on a finite geometry. On a cylinder / strip of transverse size $\beta$ the Hamiltonian
that generates translations along it is, up to the sound velocity $v$, an integral of the stress
tensor, and the conformal map from the plane produces one extra, purely geometric term — the Casimir
anomaly proportional to $c$. For our boundary (strip) geometry this is the paper's Eq. (18),

$$ \hat H \;=\; \frac{\pi}{\beta}\Big(L_0 - \frac{c}{24}\Big). \tag{C4}$$

Read it slowly, because it is the crux:

- the $-\tfrac{c}{24}$ is the **Casimir/anomaly** piece — a universal ground-state offset set *only*
  by $c$ and the size $\beta$. Isolate it and you have measured $c$.
- the $L_0$ piece contributes the **operator dimensions** $x$ on top of that offset — the excited
  levels, spaced by the $x$'s of §4.

Since the transfer matrix $T$ is essentially $e^{-\delta t\,\hat H}$ (it advances the system by one
step along the strip, §8), Eq. (C4) hands us the *entire spectrum of $T$* in one shot: a dominant
eigenvalue whose size encodes the Casimir/$c$ term, and a tower of subleading eigenvalues whose
*ratios and phases* encode the operator dimensions $x$. Concretely the paper writes the transfer
operator as (Eq. 19/2)

$$ T \;=\; \exp\!\Big[-\big(-\kappa + \pi L_0\big)\cdot(\text{geometry})\Big], \qquad \kappa = -\frac{\pi c\,\delta t}{24}. \tag{C5}$$

**This is where the campaign's two "routes" come from.** Route 2 (eigenvalues) fits the $\kappa$
(hence $c$) and the $L_0$ dimensions directly out of the leading eigenvalues — that is Eqs. (3) and
(4) below. Route 1 (entropy) instead uses the *twist-field* operators (§6, §8), whose dimensions also
carry $c$, to read $c$ off the entanglement. Same $c$, two independent windows onto it.

## 6. Entanglement entropy in a CFT, and the twist-field trick

Why does *entanglement* know about $c$? Because computing it forces you to evaluate a correlation
function of special operators whose dimension is built from $c$.

The standard (spatial) story first (Calabrese–Cardy). To get the entanglement entropy of an interval
$A$ you need $\mathrm{Tr}\,\rho_A^{\,n}$ (the $n$-th **Rényi/Tsallis** moment), then continue $n\to 1$
to reach the von Neumann entropy. The replica trick computes $\mathrm{Tr}\,\rho_A^n$ as a path
integral on $n$ copies of the plane sewn cyclically along $A$. That sewn geometry is equivalent, on a
*single* plane, to inserting a pair of **twist operators** $\sigma_n$ at the two endpoints of $A$ —
and the twist operator is itself a primary, with the beautiful dimension

$$ \Delta_n \;=\; \frac{c}{12}\Big(n - \frac{1}{n}\Big). \tag{C6}$$

Its two-point function (a power law, Eq. C2) immediately gives $\mathrm{Tr}\,\rho_A^n \sim \ell^{-4\Delta_n}$
for an interval of length $\ell$, and hence the famous logarithm

$$ S_n \;=\; \frac{c}{6}\Big(1 + \frac1n\Big)\log\ell + \text{const}, \qquad
   S_{n\to1} = \frac{c}{3}\log\ell + \text{const}. \tag{C7}$$

Two things to carry forward:

- **The entropy is a logarithm whose slope is $\propto c$.** Measuring the slope measures $c$. This is
  the entire logic of Route 1, and of notebook 4's equilibrium fits.
- **The Rényi order matters through a known factor** $(1+\tfrac1n)$. For a *boundary/half-line* cut
  (our case) the coefficient is halved, giving the family $\tfrac{c}{12}\tfrac{n+1}{n}$: this is
  $\tfrac{c}{6}$ at $n\to1$ (von Neumann) and $\tfrac{c}{8}$ at $n=2$ (Rényi-2). That is exactly why
  the thesis uses coefficient $c/6$ for the $n\to1$ entropy and $c/8$ for `gen_renyi2` — *not* $c/3$.
  (Getting this factor wrong was a real early-campaign bug; see CLAUDE.md §12–13.)

## 7. The quench as a boundary state, and the strip geometry

Now specialise to *our* problem: the **Loschmidt echo** $\mathcal L(t)=\langle\psi_0|e^{-i\hat H t}|\psi_0\rangle$,
the amplitude to return to the initial product state $|\psi_0\rangle=|X^+\rangle^{\otimes N}$ after
evolving with the *critical* Hamiltonian.

Here is the conformal reading of it. Write $\mathcal L$ as a Euclidean path integral (via the standard
$t\to -i\tau$ rotation, Appendix A of the paper). The two ends of the evolution are the initial and
final copies of $|\psi_0\rangle$. In CFT language a state you can use to *terminate* a path integral is
a **conformal boundary state** $|b\rangle$, and the special ones (**Cardy states**) correspond to
consistent boundary conditions. The upshot:

> **The initial product state acts as a conformally-invariant boundary condition.** The echo is the
> partition function of the CFT on an **infinitely long strip** whose two edges carry the boundary
> condition $|b\rangle$, and whose width is the (imaginary) evolution time.

Two ingredients decorate this picture and both appear all over the code:

- **The UV regulator $\beta_0$.** A sharp product state is not smooth enough to be a clean boundary
  state — it has structure all the way down to the lattice scale. We fix this by pre-evolving in
  *imaginary* time for a short amount $\beta_0$, which "cools" the state into a smooth conformal
  boundary state. On the lattice this is `nbeta` $=2\beta_0/\delta t$ imaginary-time sites split
  between the two ends. The strip width is then $T + 2\beta_0$, and CFT predictions are obtained at
  small $\beta_0$. *This is why every production run uses `nbeta ≥ 4`, and why `nbeta=0` gave dirty
  boundaries in the early campaign.*

- **Which boundary condition? free vs fixed.** Different initial states are different Cardy states,
  with different **boundary operator content**. The lightest boundary operator has a **boundary
  scaling dimension** $x_1$, and it controls how fast finite-size corrections decay. For the Ising
  CFT, the disordered (**free**) boundary has $x_1 = \tfrac12$ and converges fast; the ordered
  (**fixed**) boundary has $x_1 = 2$ and converges slowly. In our $\sigma_x$-field convention the
  field-polarised $|X^+\rangle$ is the *free* boundary ($x_1=\tfrac12$), which is why the thesis uses
  it and why measuring $x_1\approx\tfrac12$ (notebook 7) is a positive check that we have the boundary
  physics right. Measuring $x_1$ is Eq. (4) below.

## 8. The transverse trick: time becomes space, and "temporal" entropy

Everything so far was standard CFT. The move that defines *this* method is to read the strip
**sideways**.

Recall §2's remark: in Euclidean CFT the two coordinates are interchangeable. So instead of treating
the strip's long direction as "(imaginary) time" and quantising along space, we do the opposite: we
build a **transfer matrix $T$ that advances along the *temporal* direction**, acting on states that
live on a *time-like* slice. These states are the **temporal matrix product states** (tMPS) $|R\rangle$,
$\langle L|$ — the dominant right/left eigenvectors of $T$. This is the "$90^\circ$ rotation" of the
tensor network: what was a virtual bond becomes a physical time index, and vice versa (the mechanics
are in notebooks 2–3 and `blockpm_methods`).

Because it is the *same* CFT seen along the other axis, all the machinery of §4–§6 applies verbatim,
just with space and time swapped:

- **Eigenvalues of $T$ = operator content**, exactly as in §5. The leading eigenvalue's modulus gives
  the intensive echo $\ell = -|\lambda_0|$; its subleading partners give the operator dimensions.
- **Entanglement of $|R\rangle$/$\langle L|$ = a "temporal" entanglement.** But there is a twist
  (literally): the natural object is not a density matrix $|R\rangle\langle R|$ but the **reduced
  *transition* matrix (RTM)** built from *both* dominant vectors (paper Eq. 5),

$$ \mathcal T^{L|R} \;=\; \frac{|R\rangle\langle L|}{\langle L|R\rangle}, \qquad
   \text{(non-Hermitian, unit trace)}. \tag{C8}$$

  Its **generalized temporal entropies** $S^{\mathrm{gen}}_n = \tfrac{1}{1-n}\log\mathrm{Tr}\,(\mathcal T^{L|R})^n$
  are again computed by a twist-field two-point function — but now for a **time-like** separation. That
  is the $\sin$ (not $\sinh$) form, paper Eq. (20): $\langle\phi\phi\rangle\propto[\sin\tfrac{\pi}{T}(u_1-u_2)]^{-2x}$.
  Plugging the twist dimension (C6) into it and continuing $n\to1$ gives the master formula of the whole
  approach, the strip generalized entropy (paper Eq. 6),

$$ \boxed{\,S^{\mathrm{gen}}(t) \;=\; s_0 \;+\; \frac{i\pi c}{12} \;+\; \frac{c}{6}\,
   \log\!\Big[\frac{2T}{\pi}\sin\frac{\pi t}{T}\Big]\,}. \tag{C9}$$

The most striking physical fact hides in the words "**time-like**": because the separation is
temporal, the entropy grows only **logarithmically** in $T$ (the $\log$ in C9), *not linearly* as an
ordinary spatial entanglement entropy would. Logarithmic growth = polynomially-bounded bond
dimension = the reason the transverse contraction is efficient at all. The **entanglement barrier**
of the thesis is precisely the onset of the regime where this nice logarithm is about to be spoiled.

## 9. Decoder ring: reading the four formulas the thesis actually fits

Everything converges here. These are the formulas you keep seeing; now each piece has a home.

**(i) The transfer matrix, paper Eq. (2).**
$$ T = \exp\!\Big[-i\Big(-\tfrac{\kappa}{Tv} + \tfrac{\pi L_0}{Tv}\Big)\Big(1 + \tfrac{2i\beta_0}{T}\Big)\Big],
   \qquad \kappa = -\frac{\pi c\,\delta t}{24}. $$
*Reading:* it is just Eq. (C4)/(C5) exponentiated. $\kappa$ carries the **Casimir/$c$** term; $L_0$
carries the **operator dimensions**; $v$ is the non-universal sound velocity; the $(1+2i\beta_0/T)$
factor is the finite-$\beta_0$ cooling of §7. As $T\to\infty$ the eigenvalues $\{\tau_i\}$ approach the
**unit circle** — the *emergent dual unitarity* of §10. We track $\lambda_i = \log(-\tau_i)$.

**(ii) Central charge from the leading eigenvalue, Eq. (3).**
$$ \frac{\mathrm{Im}\,\lambda_0}{T} = a_0 + \frac{\kappa}{T^2} + \frac{a_4}{T^4} + \dots
   \;\;\Rightarrow\;\; c = \frac{24\,v\,|\kappa|}{\pi\,\delta t}. $$
*Reading:* isolate the universal $1/T^2$ Casimir coefficient $\kappa$ from the leading eigenvalue's
imaginary part, convert to $c$. This is **Route 2** for $c$ (used on Ising/Alcaraz in NB6/NB7). It
needs $\beta_0>0$ and large $T$ because the tiny $\kappa/T^2$ sits under a big non-universal constant.

**(iii) Boundary exponent from the first gap, Eq. (4).**
$$ \mathrm{Im}(\lambda_1 - \lambda_0) = -\frac{\pi x_1}{v\,T}. $$
*Reading:* the *gap* between the leading and first-subleading eigenvalue is the lightest **boundary
operator dimension** $x_1$ of §7. Getting $x_1\approx\tfrac12$ confirms the free (disordered) boundary
condition — a check that $|X^+\rangle$ is the boundary state we think it is.

**(iv) The generalized entropy chord, Eq. (6) $=$ (C9).**
$$ \mathrm{Re}\,S^{\mathrm{gen}}(t) = s_0 + \frac{c}{6}\log\!\Big[\tfrac{2T}{\pi}\sin\tfrac{\pi t}{T}\Big],
   \qquad \mathrm{Im}\,S^{\mathrm{gen}} = \frac{\pi c}{12}. $$
*Reading:* this is **Route 1**. The real part is a **chord**: plotted against $\log[\tfrac{2T}{\pi}\sin\tfrac{\pi t}{T}]$
it is a straight line of slope $\tfrac{c}{6}$ (or $\tfrac{c}{8}$ for `gen_renyi2` — the $n=2$ member,
§6). The imaginary part is a flat plateau at $\tfrac{\pi c}{12}$ (for $c=\tfrac12$: $\tfrac{\pi}{24}\approx0.131$)
— an independent read of $c$ that needs no fit at all. The "conformal dome" you see in every entropy
plot is this chord: it rises to a peak at the midpoint $t=T/2$ (where $\sin=1$) and is symmetric.

## 10. Emergent dual unitarity — the physics behind the numerical wall

One phrase recurs across the method and it deserves its own picture. From Eq. (2), as $T$ grows the
eigenvalues $\tau_i$ of the temporal transfer matrix migrate onto the **unit circle**: $|\tau_i|\to1$
with only their *phases* (the operator dimensions) distinguishing them. A matrix whose spectrum lies
on the unit circle is **unitary**. So in the long-time limit the temporal transfer matrix becomes
unitary — a property called **dual unitarity** (unitary in *both* the space and time directions). It
is a genuine, physical hallmark of a clean quench to a conformal point, and the thesis confirms it by
tracing $\lambda_0(T)$ around a circle in the complex plane (the "$\lambda_0$ circle" plots, NB6/NB7).

But dual unitarity is a double-edged sword, and this is the bridge to the method's limits:

- **Physically** it is exactly what we want — the fingerprint of emergent conformal/unitary structure.
- **Numerically** it is a disaster for a power method. If *all* leading eigenvalues have $|\tau_i|\to1$,
  the spectral gap ratio $|\tau_1/\tau_0|\to1$: the transfer matrix stops separating the dominant
  eigenvector from its neighbours. The single-vector power method stalls; the **block** method
  (`blockpm_methods`) survives longer by tracking the whole near-degenerate cluster at once; and
  eventually even the block method's *eigenvectors* (though not its eigenvalues) become ill-defined —
  the near-**exceptional-point** collapse studied in notebook 13.

So "emergent dual unitarity", "the entanglement barrier", "the closing spectral gap", and "the wall"
are four names for one thing seen from four angles: *the CFT working perfectly is what makes the
classical simulation eventually fail.* Characterising exactly where and why that failure sets in — as
a function of $c$ (bigger $c$ walls sooner), of frustration, and of the quench's symmetry — is a
central result of the thesis.
